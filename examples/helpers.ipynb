```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Chessboard settings
squares_x = 9
squares_y = 7
square_size = 25

# Power-of-two image dimensions
image_width = 256
image_height = 256

# Chessboard dimensions
board_width = squares_x * square_size   # 225 px
board_height = squares_y * square_size  # 150 px

# Center the board precisely
start_x = (image_width - board_width) // 2   # 143
start_y = (image_height - board_height) // 2 # 53

# White background
image = np.full(
    (image_height, image_width),
    255,
    dtype=np.uint8
)

# Draw the chessboard
for y in range(squares_y):
    for x in range(squares_x):
        if (x + y) % 2 == 0:
            x1 = start_x + x * square_size
            y1 = start_y + y * square_size

            image[y1:y1 + square_size,
                  x1:x1 + square_size] = 0

# Display
plt.figure(figsize=(12, 6))
plt.imshow(image, cmap="gray", vmin=0, vmax=255)
plt.axis("off")
plt.show()

# Save
filename = "chessboard_9x7_256x256.png"
cv2.imwrite(filename, image)

print(f"Saved: {filename}")
print(f"Image: {image_width} x {image_height} px")
print(f"Board: {board_width} x {board_height} px")
print(f"Margins: left={start_x}, right={image_width-start_x-board_width}, "
      f"top={start_y}, bottom={image_height-start_y-board_height}")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Configuration
marker_id = 5
marker_size = 100
image_size = 128

# Original ArUco dictionary
aruco_dict = cv2.aruco.getPredefinedDictionary(
    cv2.aruco.DICT_ARUCO_ORIGINAL
)

# Generate marker (100 x 100 pixels)
marker = cv2.aruco.generateImageMarker(
    aruco_dict,
    marker_id,
    marker_size
)

# Create white 128 x 128 image
image = np.full(
    (image_size, image_size),
    255,
    dtype=np.uint8
)

# Center the marker
start = (image_size - marker_size) // 2  # 14 pixels

image[
    start:start + marker_size,
    start:start + marker_size
] = marker

# Display
plt.figure(figsize=(5, 5))
plt.imshow(image, cmap="gray", vmin=0, vmax=255)
plt.axis("off")
plt.show()

# Save
filename = "aruco_original_id5_128x128.png"
cv2.imwrite(filename, image)

print(f"Saved: {filename}")
print(f"Image size: {image.shape[1]} x {image.shape[0]} pixels")
print(f"Marker size: {marker_size} x {marker_size} pixels")
print(f"White border: {start} pixels on each side")
```
